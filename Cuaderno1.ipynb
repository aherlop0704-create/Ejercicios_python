```python
frase = input("Escribe una frase: ")

print(f"Frase en mayuscula: {frase.upper()}")
print(f"Frase en minuscula: {frase.lower()}")
print(f"Caracteres: {len(frase)}")
print(f"La letra a aparece: {frase.lower().count("a")}")
print(f"Frase sin espacios ni al principio ni al final: {frase.strip()}")
print(f"Primera palabra: {frase.split()[0]}")
print(f"Frase al reves: {frase[::-1]}")
```
